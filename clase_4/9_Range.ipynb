# **Range()**

Retorna una lista de elementos desde 0 y hasta el parámetro menos uno que se envíe dentro de los paréntesis.

Retorna solo para números enteros.

```python
# Lista de 0 a 9
range(10)
```

```python
list(range(10))
```

```python
list(range(10,20,3))
```

```python
# Lista de 10 a 19, cada 2 elementos
list(range(10,20,2))
```

```python
# Lista de 0 a 4
range(5)
```

```python
list(range(5))
```

```python
for i in range(5):
    print(i)
```
